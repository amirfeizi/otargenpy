# otargenpy — Quick Start Tutorial

This notebook demonstrates the core query and plotting functions in **otargenpy**, the Python interface to the [Open Targets Platform](https://platform.opentargets.org) GraphQL API.

Every function returns a **pandas DataFrame** ready for analysis.

```python
# Install (uncomment if needed)
# !pip install git+https://github.com/amirfeizi/otargenpy.git
```

```python
import otargenpy as ot
print(f"otargenpy version: {ot.__version__}")
```

---
## 1. Drug Queries — Adverse Events

Retrieve and visualize adverse events for **imatinib** (CHEMBL941), a well-known tyrosine kinase inhibitor.

```python
ae = ot.adverse_events_query("CHEMBL941", size=30)
print(f"Shape: {ae.shape}")
ae.head()
```

```python
ot.plot_adverse_events(ae)
```

---
## 2. Drug Indications by Clinical Stage

Which diseases is imatinib approved or in clinical trials for?

```python
ind = ot.indications_query("CHEMBL941")
print(f"Shape: {ind.shape}")
ind[["disease.name", "maxClinicalStage"]].head(10)
```

```python
ot.plot_indications(ind)
```

---
## 3. Gene Queries — Protein Interactions

Retrieve protein–protein interactions for **TP53** (ENSG00000141510) from the IntAct database.

```python
inter = ot.interactions_query("ENSG00000141510", source_database="intact", size=25)
print(f"Shape: {inter.shape}")
inter[["targetA.approvedSymbol", "targetB.approvedSymbol", "score"]].head()
```

```python
ot.plot_interactions(inter)
```

---
## 4. Gene + Disease Evidence — ChEMBL

Retrieve ChEMBL clinical evidence linking **PARP1** to **breast cancer**.

```python
ev = ot.chembl_query("ENSG00000143799", "EFO_0000305")
print(f"Shape: {ev.shape}")
ev.head()
```

---
## 5. Genetics — Locus-to-Gene Predictions

L2G scores rank candidate causal genes for a GWAS locus.

```python
l2g = ot.locus2gene_query("fa375739ca2a6b825ce5cc69d117e84b")
print(f"Shape: {l2g.shape}")
l2g[["target.approvedSymbol", "score"]].head(10)
```

```python
ot.plot_l2g(l2g)
```

---
## 6. Genetics — GWAS Colocalisation

Which other GWAS traits share a causal variant with this locus?

```python
coloc = ot.gwas_colocalisation("5a86bfd40d2ebecf6ce97bbe8a737512")
print(f"Shape: {coloc.shape}")
coloc.head()
```

```python
ot.plot_colocalisation(coloc)
```

---
## 7. More Queries — Explore on Your Own

otargenpy has **40 functions** covering drugs, genes, diseases, variants, and genetics.
Here are a few more to try:

```python
# Cancer hallmarks for TP53
hall = ot.hallmarks_query("ENSG00000141510")
display(hall.head())

# Genetic constraint (pLI, LOEUF) for BRCA1
gc = ot.genetic_constraint_query("ENSG00000012048")
display(gc)

# Safety liabilities for EGFR
safe = ot.safety_query("ENSG00000146648")
display(safe.head())

# Variant effect predictions
vep = ot.variant_effect_predictor_query("1_154453788_C_T")
display(vep.head())
```
